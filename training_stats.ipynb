## Utility notebook to mass-generate training stat plots for models

```python
from pathlib import Path

import pandas as pd

import visualization as vis
```

```python
BASE_DIR = Path('.')
MODEL_DIR = BASE_DIR / 'trained_models'
```

```python
for subfolder in MODEL_DIR.glob('final-*/'):
    stats_file = subfolder / 'stats.csv'
    if not stats_file.exists():
        print(f'{stats_file} does not exist, skipping')
        continue
    stats = pd.read_csv(stats_file)
    print(f'Plotting training stats for {stats_file}')
    save_path_prefix = str(subfolder / 'trainingstats')
    vis.plot_training_curves(stats, log_corr=True, save_path=save_path_prefix)
```
